```python
import json
import matplotlib.pyplot as plt
```

```python
# Load both logs
with open("best/threshold_log.json") as f:
    full_data = json.load(f)

with open("best/threshold_log_small.json") as f:
    small_data = json.load(f)

# Split X/Y
full_thresh = [x["threshold"] for x in full_data]
full_f1 = [x["f1"] for x in full_data]

small_thresh = [x["threshold"] for x in small_data]
small_f1 = [x["f1"] for x in small_data]

plt.figure(figsize=(8, 5))

plt.plot(full_thresh, full_f1, label="Full Training Data", linewidth=2)
plt.plot(small_thresh, small_f1, label="10% Training Data", linestyle="--", linewidth=2)

plt.xlabel("Threshold")
plt.ylabel("Macro-F1")
plt.title("Macro-F1 vs Threshold\nFull Dataset vs 10% Dataset")
plt.legend()
plt.grid(True, linestyle=":")

plt.tight_layout()
plt.show()
```
